## Tablas de frecuencias

  <b><font size="4">Código básico:</font></b>

<p style="margin-top: 15px;">
  <b><font size="4" color="gray">Datos correspondientes a una variable cuantitativa:</font></b>
</p>


- Pocos valores distintos
- Muchos valores distintos

```r
# Creamos el vector con los datos
datos <- c(18.5, 19.2, 20.1, 21.3, 22.5, 20.8, 19.7, 21.0, 20.4, 19.9, 22.1, 21.7, 20.6, 21.2, 20.3, 22.0, 21.5, 19.5, 20.0, 21.9)
```

```r
# Verificamos la condición
n_distintos <- length(unique(datos))
umbral <- 10  # Usualmente son 10/15

cat("La cantidad de valores distintos es de:", n_distintos, "\n")

# Si la cantidad de valores distintos es menor o igual al umbral propuesto, realizamos una tabla no agrupada, caso contrario realizamos una tabla agrupada por intervalos
if (n_distintos <= umbral) {
  cat("Como hay menos o igual a", umbral, "valores distintos, realizamos una 'Tabla no agrupada'\n\n")

  tabla_frecuencias <- tibble(xi = sort(datos)) %>%
    count(xi, name = "fi") %>%
    mutate(
      Fi = cumsum(fi),
      fri = fi / sum(fi),
      Fri = cumsum(fri),
      pi = round(fri * 100, 2),
      Pi = round(Fri * 100, 2)
    )

}else {

  cat("Como hay más de", umbral, "valores distintos, realizamos una 'Tabla agrupada por intervalos'\n\n")

  # Valor mínimo
  x_min <- min(datos)

  # Valor máximo
  x_max <- max(datos)

  # Tamaño de la muestra
  n <- length(datos)

  # Número de clases
  k <- ceiling(sqrt(n))

  # Ancho o amplitud de la clase
  A <- ceiling((x_max - x_min) / k)

  # Límite inicial (redondeado hacia abajo)
  L0 <- floor(x_min)

  # Sumamos la amplitud sucesivamente hasta superar el máximo
  cortes <- seq(from = L0, to = x_max + A, by = A)

  # Creamos la tabla de frecuencia
  tabla_frecuencias <- tibble(x = datos) %>%
    mutate(intervalo = cut(x, breaks = cortes, right = FALSE, include.lowest = TRUE)) %>%
    count(intervalo, name = "fi") %>%
    mutate(
      xi = cortes[1:n()] + (A / 2),
      Fi = cumsum(fi),
      fri = fi / sum(fi),
      Fri = cumsum(fri),
      pi = round(fri * 100, 2),
      Pi = round(Fri * 100, 2)
    )
}

tabla_frecuencias
```

## Medidas descriptivas

```r
# Armamos el vector con los datos
datos <- c(12, 15, 14, 10, 18, 15, 16, 14, 19, 20, 15, 17, 13, 11, 14, 16, 22, 15, 13, 14)
```

```r
# Calculamos los descriptivos
descriptivos <- tibble(
  conteo   = length(datos),
  media    = round(mean(datos, na.rm = TRUE), 2),
  varianza = round(var(datos, na.rm = TRUE), 2),
  desvio   = round(sd(datos, na.rm = TRUE), 2),
  moda     = paste(mfv(datos), collapse = ", "),
  minimo   = round(min(datos, na.rm = TRUE), 2),
  maximo   = round(max(datos, na.rm = TRUE), 2),
  IQR      = round(IQR(datos, na.rm = TRUE), 2),
  rango    = maximo - minimo,
  mediana  = round(median(datos, na.rm = TRUE), 2),
  Q1       = round(quantile(datos, 0.25, na.rm = TRUE), 2),
  Q3       = round(quantile(datos, 0.75, na.rm = TRUE), 2),
  sup.     = round((Q3 + 1.5 * IQR), 2),
  inf.     = round((Q1 - 1.5 * IQR), 2),
  CV       = round(sd(datos, na.rm = TRUE) / mean(datos, na.rm = TRUE) * 100, 2)
)

descriptivos
```
